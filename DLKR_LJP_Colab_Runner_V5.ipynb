# ⚖️ DLKR-LJP V5.1 Flagship: Top-Tier State-of-the-Art Legal Judgment Prediction
### GraphRAG-Augmented Dual-Code Jurisprudence (IPC 1860 & BNS 2023)

**State-of-the-Art Architectural Pillars:**
1. **GraphRAG Statutory Knowledge Graph**: Directed Acyclic Graph (DAG) traversing statutory provisions, Section 300 Exceptions 1–5, and Supreme Court precedent ratios (*Pulicherla*, *Virsa Singh*, *Ghapoo Yadav*, *Arnesh Kumar*, *State of Haryana v. Bhajan Lal*).
2. **Attention-Weighted Mean Pooling Encoder**: 768-dim InLegalBERT representation pooled across all 512 tokens — resolving representation collapse and eliminating the single `[CLS]` bottleneck.
3. **Cost-Sensitive Asymmetric Loss**: Penalizes false negatives on fatal/heinous crimes (Murder, Culpable Homicide, Rape) up to 10x more than minor theft, preventing fatal-to-hurt undercharging.
4. **Anti-Leakage Rhetorical Fact Extractor**: Extracts pure pre-trial incident facts with scoped negation handling, discarding circular post-trial judicial conclusions.
5. **Temporal Statutory Regime Transition**: Automated temporal detection routing offences pre-July 1, 2024 to IPC 1860 and post-July 1, 2024 to BNS 2023.
6. **Judicial Consistency Verifier (JCV Module)**: Strict symbolic auditor enforcing statutory penalty ceilings, Arnesh Kumar 7-year procedural floors, and Section 482 CrPC / Section 528 BNSS quashing.
