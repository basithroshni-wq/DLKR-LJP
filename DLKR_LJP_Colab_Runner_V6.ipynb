## 🔍 Step 2: Instant Single-Case Judicial Adjudication
Test an incident narrative and inspect the complete 7-Point Traceable Evidence & Statutory Graph Traversal report.

```python
import os
from dlkr_ljp_v5_final import get_pipeline, SAMPLE_KAKKANAD_EXCEPTION4


pipeline = get_pipeline()

# Adjudicate test incident (Sudden quarrel -> Culpable Homicide vs Murder)
case_narrative = SAMPLE_KAKKANAD_EXCEPTION4
result = pipeline.run(case_narrative)
preds = result['predictions']
graph = preds.get('statutory_graph_audit', {})
jcv = preds.get('judicial_consistency_audit', {})

print("=" * 75)
print("⚖️ DLKR-LJP V6 ADJUDICATION & STATUTORY GRAPHRAG REPORT")
print("=" * 75)
if preds.get('head1_articles'):
    print(f"📌 Primary Law Article : {preds['head1_articles'][0]['display_name']}")
    print(f"   Confidence Score    : {preds['head1_articles'][0]['probability']:.1%}")
if preds.get('head2_charges'):
    print(f"📌 Criminal Charge     : {preds['head2_charges'][0]['charge_name']}")
if preds.get('primary_penalty'):
    print(f"📌 Penalty Bracket     : {preds['primary_penalty']['tier']}")
if graph.get('traversal_path'):
    print(f"📌 Graph Traversal     : {' -> '.join(graph['traversal_path'])}")
if graph.get('applied_exception'):
    print(f"📌 Applied Exception   : {graph['applied_exception']}")
if jcv.get('audit_status'):
    print(f"📌 JCV Consistency     : {jcv['audit_status']} (JCI Score: {preds.get('judicial_consistency_index', 100)}%)")
print("=" * 75)
print("\n--- FULL TRACEABLE EVIDENCE AUDIT ---")
print(result.get('traceable_evidence', ''))
```
