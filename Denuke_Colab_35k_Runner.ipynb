**Core Architecture Highlights:**
1. **35,000 Real Indian Court Judgments**: Trained on the Indian Legal Dataset Corpus (ILDC).
2. **Anti-Leakage Rhetorical Role Extraction**: Isolates pure pre-trial incident facts; eliminates judicial outcome leakage.

4. **Attention Mean-Pooled InLegalBERT**: Full 512-token span representation.
5. **Cost-Sensitive Asymmetric Loss**: Penalizes homicide false negatives 10x higher.
